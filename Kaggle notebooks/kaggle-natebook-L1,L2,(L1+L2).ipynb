```python
#step 20
import os

required_files = [
    "data/iemocap/IEMOCAP_features.pkl",
    "data/iemocap/iemocap_emotion_features_roberta.pkl",
    "data/iemocap/iemocap_emotion_semantic_features_roberta.pkl",
    "data/meld/meld_emotion_semantic_features_roberta.pkl",
]

print("Checking required ECERC feature files...\n")

all_present = True

for path in required_files:
    exists = os.path.isfile(path)

    if exists:
        size_mb = os.path.getsize(path) / (1024 ** 2)
        print(f"✅ {path}")
        print(f"   Size: {size_mb:.2f} MB")
    else:
        print(f"❌ MISSING: {path}")
        all_present = False

print("\n--------------------------------")

if all_present:
    print("✅ ALL FOUR REQUIRED DATASET FILES ARE PRESENT.")
else:
    print("❌ ONE OR MORE DATASET FILES ARE MISSING.")
```
