```python
# CELL 3 — Extract exact per-class results + confusion matrices

import re
import json
import numpy as np
import pandas as pd
from pathlib import Path

def parse_log(log_path, dataset, method, labels):
    text = Path(log_path).read_text(errors="ignore")

    start = text.find("Per-class test report")
    if start == -1:
        raise ValueError(f"Report not found: {log_path}")

    end = text.find("Validation scores", start)
    if end == -1:
        end = len(text)

    report_text = text[start:end]

    # ---------- Per-class metrics ----------
    rows = []

    for line in report_text.splitlines():
        m = re.match(
            r"\s*([A-Za-z_]+)\s+"
            r"([0-9.]+)\s+"
            r"([0-9.]+)\s+"
            r"([0-9.]+)\s+"
            r"([0-9]+)\s*$",
            line
        )

        if m and m.group(1) in labels:
            rows.append({
                "dataset": dataset,
                "method": method,
                "class": m.group(1),
                "precision": float(m.group(2)),
                "recall": float(m.group(3)),
                "f1": float(m.group(4)),
                "support": int(m.group(5)),
            })

    # ---------- NumPy-style confusion matrix ----------
    matrix_lines = []

    for line in report_text.splitlines():
        line = line.strip()

        if line.startswith("[") and line.endswith("]"):
            nums = re.findall(r"\d+", line)

            # Ignore the outer "[]" unless this is a real matrix row
            if len(nums) >= len(labels):
                matrix_lines.append([int(x) for x in nums])

    # Take the last N rows corresponding to the confusion matrix
    n = len(labels)

    if len(matrix_lines) < n:
        raise ValueError(
            f"Could not find {n} confusion-matrix rows in {log_path}. "
            f"Found {len(matrix_lines)}."
        )

    matrix = np.array(matrix_lines[-n:], dtype=int)

    if matrix.shape != (n, n):
        raise ValueError(
            f"Bad confusion matrix shape {matrix.shape} "
            f"for {dataset}/{method}; expected {(n, n)}"
        )

    return rows, matrix


log_specs = [
    (
        "MELD",
        "baseline",
        ROOT / "tuning_results/MELD/logs/baseline_s0.txt",
        ["neu", "sur", "fea", "sad", "joy", "dis", "ang"]
    ),
    (
        "MELD",
        "L2",
        ROOT / "tuning_results/MELD/logs/L2_e0.02_m0.5_s0.txt",
        ["neu", "sur", "fea", "sad", "joy", "dis", "ang"]
    ),
    (
        "MELD",
        "ALL",
        ROOT / "tuning_results/MELD/logs/final_ALL_s0.txt",
        ["neu", "sur", "fea", "sad", "joy", "dis", "ang"]
    ),
    (
        "IEMOCAP",
        "baseline",
        ROOT / "tuning_results/IEMOCAP/logs/baseline_s2007.txt",
        ["hap", "sad", "neu", "ang", "exc", "fru"]
    ),
    (
        "IEMOCAP",
        "L2",
        ROOT / "tuning_results/IEMOCAP/logs/L2_e0.1_m0.9_s2007.txt",
        ["hap", "sad", "neu", "ang", "exc", "fru"]
    ),
    (
        "IEMOCAP",
        "ALL",
        ROOT / "tuning_results/IEMOCAP/logs/final_ALL_s2007.txt",
        ["hap", "sad", "neu", "ang", "exc", "fru"]
    ),
]


per_class_rows = []
matrices = {}

for dataset, method, path, labels in log_specs:
    print(f"Reading {dataset} / {method} ...")

    rows, matrix = parse_log(path, dataset, method, labels)

    per_class_rows.extend(rows)

    matrices[f"{dataset}_{method}"] = {
        "labels": labels,
        "matrix": matrix.tolist()
    }

per_class_df = pd.DataFrame(per_class_rows)

per_class_df.to_csv(
    OUT / "final_per_class_results.csv",
    index=False
)

with open(OUT / "confusion_matrices.json", "w") as f:
    json.dump(matrices, f, indent=2)

print("\n========== PER-CLASS RESULTS ==========")
print(per_class_df.to_string(index=False))

print("\n========== CONFUSION MATRICES ==========")

for key, info in matrices.items():
    print(f"\n{key}")
    print(np.array(info["matrix"]))

print("\nSaved:")
print(OUT / "final_per_class_results.csv")
print(OUT / "confusion_matrices.json")
```

```python
def plot_confusion(dataset, method):
    info = matrices[f"{dataset}_{method}"]
    labels = info["labels"]
    matrix = np.array(info["matrix"])

    fig, ax = plt.subplots(figsize=(7, 6))
    ax.imshow(matrix)

    ax.set_title(f"{dataset} — {method}")
    ax.set_xlabel("Predicted")
    ax.set_ylabel("True")
    ax.set_xticks(range(len(labels)))
    ax.set_yticks(range(len(labels)))
    ax.set_xticklabels(labels)
    ax.set_yticklabels(labels)

    threshold = matrix.max() / 2

    for i in range(matrix.shape[0]):
        for j in range(matrix.shape[1]):
            ax.text(
                j, i, str(matrix[i, j]),
                ha="center",
                va="center"
            )

    plt.tight_layout()

    path = OUT / f"confusion_{dataset}_{method}.png"
    plt.savefig(path, dpi=200, bbox_inches="tight")
    plt.close()

    return path


for dataset in ["MELD", "IEMOCAP"]:
    for method in ["baseline", "L2", "ALL"]:
        plot_confusion(dataset, method)


# Pairwise confusions explicitly relevant to your SOP
def pair_sum(matrix, labels, a, b):
    i = labels.index(a)
    j = labels.index(b)
    return int(matrix[i, j] + matrix[j, i])


pair_rows = []

# IEMOCAP: Happy/Excited and Angry/Frustrated
for method in ["baseline", "L2", "ALL"]:
    info = matrices[f"IEMOCAP_{method}"]
    m = np.array(info["matrix"])
    labels = info["labels"]

    pair_rows.append([
        "IEMOCAP", method, "Happy ↔ Excited",
        pair_sum(m, labels, "hap", "exc")
    ])

    pair_rows.append([
        "IEMOCAP", method, "Angry ↔ Frustrated",
        pair_sum(m, labels, "ang", "fru")
    ])

# MELD: Disgust/Angry
for method in ["baseline", "L2", "ALL"]:
    info = matrices[f"MELD_{method}"]
    m = np.array(info["matrix"])
    labels = info["labels"]

    pair_rows.append([
        "MELD", method, "Disgust ↔ Angry",
        pair_sum(m, labels, "dis", "ang")
    ])

pair_df = pd.DataFrame(
    pair_rows,
    columns=["dataset", "method", "pair", "two_way_confusions"]
)

pair_df.to_csv(
    OUT / "confusable_pair_audit.csv",
    index=False
)

print(pair_df.to_string(index=False))
```

```python
span_frames = {}

for dataset in ["MELD", "IEMOCAP"]:
    sweep_path = ROOT / f"{dataset}/span_curve_final/span_sweep_{dataset}.csv"
    learned_path = ROOT / f"{dataset}/span_curve_final/learned_span_curve_{dataset}.json"

    sweep = pd.read_csv(sweep_path)

    sweep["numeric_span"] = pd.to_numeric(
        sweep["span"], errors="coerce"
    )

    sweep = sweep[sweep["numeric_span"].notna()].copy()
    sweep["dataset"] = dataset
    sweep["source"] = "fixed_span_sweep"

    learned = pd.read_json(learned_path)

    learned["numeric_span"] = pd.to_numeric(
        learned["x"], errors="coerce"
    )

    learned = learned[learned["numeric_span"].notna()].copy()
    learned["dataset"] = dataset
    learned["source"] = "learned_span"

    combined = pd.concat([
        sweep[[
            "dataset", "source",
            "numeric_span", "acc",
            "weighted_f1", "ece",
            "nll", "avg_conf"
        ]],
        learned[[
            "dataset", "source",
            "numeric_span", "acc",
            "weighted_f1", "ece",
            "nll", "avg_conf"
        ]]
    ], ignore_index=True)

    span_frames[dataset] = combined

    combined.to_csv(
        OUT / f"final_span_results_{dataset}.csv",
        index=False
    )


# Accuracy / F1 vs span
for dataset, df in span_frames.items():

    fig, ax = plt.subplots(figsize=(9, 6))

    fixed = df[df["source"] == "fixed_span_sweep"]
    learned = df[df["source"] == "learned_span"]

    ax.plot(
        fixed["numeric_span"],
        fixed["weighted_f1"],
        marker="o",
        label="Fixed span sweep"
    )

    ax.plot(
        learned["numeric_span"],
        learned["weighted_f1"],
        marker="s",
        label="Learned span checkpoints"
    )

    ax.set_xlabel("Attention span (utterances)")
    ax.set_ylabel("Weighted F1")
    ax.set_title(f"{dataset} — Weighted F1 vs Attention Span")
    ax.legend()

    plt.tight_layout()

    plt.savefig(
        OUT / f"span_vs_weighted_f1_{dataset}.png",
        dpi=200,
        bbox_inches="tight"
    )
    plt.close()


# ECE vs span
for dataset, df in span_frames.items():

    fig, ax = plt.subplots(figsize=(9, 6))

    fixed = df[df["source"] == "fixed_span_sweep"]
    learned = df[df["source"] == "learned_span"]

    ax.plot(
        fixed["numeric_span"],
        fixed["ece"],
        marker="o",
        label="Fixed span sweep"
    )

    ax.plot(
        learned["numeric_span"],
        learned["ece"],
        marker="s",
        label="Learned span checkpoints"
    )

    ax.set_xlabel("Attention span (utterances)")
    ax.set_ylabel("ECE (%)")
    ax.set_title(f"{dataset} — ECE vs Attention Span")
    ax.legend()

    plt.tight_layout()

    plt.savefig(
        OUT / f"span_vs_ece_{dataset}.png",
        dpi=200,
        bbox_inches="tight"
    )
    plt.close()

print("Span plots generated.")
```
